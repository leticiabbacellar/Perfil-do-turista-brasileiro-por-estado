```python
import pandas as pd
```

```python
gasto_2020= pd.read_csv ("/content/drive/MyDrive/tabelas/gasto/Tabela 8464 - Gasto médio 2020.csv", sep=";", skiprows= 4 )
gasto_2021= pd.read_csv ("/content/drive/MyDrive/tabelas/gasto/Tabela 8464 - Gasto médio 2021.csv", sep=";", skiprows=4)
gasto_2023= pd.read_csv ("/content/drive/MyDrive/tabelas/gasto/Tabela 8464 - Gasto médio 2023.csv", sep=";", skiprows= 4)
gasto_2024= pd.read_csv ("/content/drive/MyDrive/tabelas/gasto/Tabela 8464 - Gasto médio 2024.csv", sep=";", skiprows=4)

gasto_2020["Ano"] = 2020
gasto_2021["Ano"] = 2021
gasto_2023["Ano"] = 2023
gasto_2024["Ano"] = 2024

gasto = pd.concat([gasto_2020, gasto_2021, gasto_2023, gasto_2024])
for df in [gasto]:
    df.rename(columns={'Grande Região e Unidade da Federação de origem': 'UF'}, inplace=True)

if 'Unnamed: 0' in gasto.columns:
  gasto.drop(columns=["Unnamed: 0"], inplace=True)
print (gasto.head())
```

```python
finalidade_2020= pd.read_csv ("/content/drive/MyDrive/tabelas/finalidade/Tabela 7204 - finalidade da viagem 2020.csv", sep= ";", skiprows=4)
finalidade_2021= pd.read_csv ("/content/drive/MyDrive/tabelas/finalidade/Tabela 7204 - finalidade da viagem 2021.csv", sep= ";", skiprows=4)
finalidade_2023= pd.read_csv ("/content/drive/MyDrive/tabelas/finalidade/Tabela 7204 - finalidade da viagem 2023.csv", sep= ";", skiprows=4)
finalidade_2024= pd.read_csv ("/content/drive/MyDrive/tabelas/finalidade/Tabela 7204 - finalidade da viagem 2024.csv", sep= ";", skiprows=4)

finalidade_2020["Ano"] = 2020
finalidade_2021["Ano"] = 2021
finalidade_2023["Ano"] = 2023
finalidade_2024["Ano"] = 2024

finalidade = pd.concat([finalidade_2020, finalidade_2021, finalidade_2023, finalidade_2024])

for df in[finalidade]:
  df.rename(columns={"Unidade da Federação": "UF"}, inplace=True)
  print (df.head())
```

```python
hospedagem_2020 = pd.read_csv ("/content/drive/MyDrive/tabelas/hospedagem/Tabela 8463 - hospedagem 2020.csv", sep= ";", skiprows= 4)
hospedagem_2021= pd.read_csv ("/content/drive/MyDrive/tabelas/hospedagem/Tabela 8463 - hospedagem 2021.csv", sep= ";", skiprows= 4)
hospedagem_2023= pd.read_csv ("/content/drive/MyDrive/tabelas/hospedagem/Tabela 8463 - hospedagem 2023.csv", sep= ";", skiprows= 4)
hospedagem_2024= pd.read_csv ("/content/drive/MyDrive/tabelas/hospedagem/Tabela 8463 - Tipo de hospedagem 2024.csv", sep= ";", skiprows= 4)

hospedagem_2020["Ano"] = 2020
hospedagem_2021["Ano"] = 2021
hospedagem_2023["Ano"] = 2023
hospedagem_2024["Ano"] = 2024

hospedagem = pd.concat([hospedagem_2020, hospedagem_2021, hospedagem_2023, hospedagem_2024])
for df in [hospedagem]:
  df.rename(columns={"Unidade da Federação": "UF"}, inplace=True)
  print (df.head())

```

```python
transporte_2020 = pd.read_csv ("/content/drive/MyDrive/tabelas/transporte/Tabela 7263 - transporte utilizado 2020.csv", sep= ";", skiprows=4)
transporte_2021= pd.read_csv ("/content/drive/MyDrive/tabelas/transporte/Tabela 7263 - transporte utilizado 2021.csv", sep =";", skiprows= 4)
transporte_2023= pd.read_csv ("/content/drive/MyDrive/tabelas/transporte/Tabela 7263 - transporte utilizado 2023.csv", sep= ";", skiprows=4)
transporte_2024= pd.read_csv (  "/content/drive/MyDrive/tabelas/transporte/Tabela 7263 - transporte utilizado 2024.csv", sep= ";", skiprows=4)

transporte_2020["Ano"] = 2020
transporte_2021["Ano"] = 2021
transporte_2023["Ano"] = 2023
transporte_2024["Ano"] = 2024

transporte = pd.concat([transporte_2020, transporte_2021, transporte_2023, transporte_2024])
for df in [transporte]:
  df.rename(columns={"Unidade da Federação": "UF"}, inplace=True)
  print (df.head())

```

```python
ocor_final_hosp= pd.merge(finalidade, hospedagem, on=['UF', "Ano"], how='outer')
print (ocor_final_hosp.head())
```

```python
ocor_final_hosp_transp= pd.merge(ocor_final_hosp, transporte, on=['UF', "Ano"], how='outer')
print (ocor_final_hosp_transp.head())
```

```python
base_perfil = pd.merge(ocor_final_hosp_transp, gasto, on=['UF', "Ano"], how='outer')

regioes_remover = ['Sudeste', 'Nordeste', 'Centro-Oeste', 'Norte', 'Sul']

numeric_cols = base_perfil.columns.drop(['UF', 'Ano'])
for col in numeric_cols:
    base_perfil[col] = pd.to_numeric(base_perfil[col], errors='coerce')

base_perfil = base_perfil[~base_perfil['UF'].isin(regioes_remover)]


base_perfil = base_perfil.fillna(0)

#base_perfil.dtypes
#base_perfil.to_csv('/content/base_perfil.csv', index=False, encoding='utf-8')
```

```python
# Total finalidade
base_perfil['Total_finalidade'] = (
    base_perfil['Pessoal'] +
    base_perfil['Profissional']
)

# Proporções
base_perfil['prop_pessoal'] = base_perfil['Pessoal'] / base_perfil['Total_finalidade']
base_perfil['prop_profissional'] = base_perfil['Profissional'] / base_perfil['Total_finalidade']


base_perfil['Total_finalidade'] > 0
```

```python
# Total hospedagem
base_perfil['Total_hospedagem'] = (
    base_perfil['Hotel, resort ou flat'] +
    base_perfil['Pousada'] +
    base_perfil['Casa de amigo ou parente'] +
    base_perfil['Imóvel próprio'] +
    base_perfil['Imóvel por temporada ou AirBnB'] +
    base_perfil['Outro_x']
)

# Proporções
base_perfil['prop_hotel'] = base_perfil['Hotel, resort ou flat'] / base_perfil['Total_hospedagem']
base_perfil['prop_pousada'] = base_perfil['Pousada'] / base_perfil['Total_hospedagem']
base_perfil['prop_casa'] = base_perfil['Casa de amigo ou parente'] / base_perfil['Total_hospedagem']
base_perfil['prop_imovel_proprio'] = base_perfil['Imóvel próprio'] / base_perfil['Total_hospedagem']
base_perfil['prop_airbnb'] = base_perfil['Imóvel por temporada ou AirBnB'] / base_perfil['Total_hospedagem']
base_perfil['prop_outro_hosp'] = base_perfil['Outro_x'] / base_perfil['Total_hospedagem']

base_perfil['Total_hospedagem'] > 0
```

```python
# Total transporte
base_perfil['Total_transporte'] = (
    base_perfil['Avião'] +
    base_perfil['Carro particular ou de empresa'] +
    base_perfil['Motocicleta'] +
    base_perfil['Ônibus de excursão, fretado ou turismo'] +
    base_perfil['Ônibus de linha'] +
    base_perfil['Van ou perueiro'] +
    base_perfil['Outro_y']
)

# Proporções
base_perfil['prop_aviao'] = base_perfil['Avião'] / base_perfil['Total_transporte']
base_perfil['prop_carro'] = base_perfil['Carro particular ou de empresa'] / base_perfil['Total_transporte']
base_perfil['prop_moto'] = base_perfil['Motocicleta'] / base_perfil['Total_transporte']
base_perfil['prop_onibus_exc'] = base_perfil['Ônibus de excursão, fretado ou turismo'] / base_perfil['Total_transporte']
base_perfil['prop_onibus_linha'] = base_perfil['Ônibus de linha'] / base_perfil['Total_transporte']
base_perfil['prop_van'] = base_perfil['Van ou perueiro'] / base_perfil['Total_transporte']
base_perfil['prop_outro_transp'] = base_perfil['Outro_y'] / base_perfil['Total_transporte']

base_perfil['Total_transporte'] > 0
```

```python
# Total distribuição de gasto médio
base_perfil['Total_gasto'] = (
    base_perfil['Menos de 1/2 salário mínimo'] +
    base_perfil['1/2 a menos de 1 salário mínimo'] +
    base_perfil['1 a menos de 2 salários mínimos'] +
    base_perfil['2 a menos de 4 salários mínimos'] +
    base_perfil['4 ou mais salários mínimos']
)

base_perfil['prop_gasto_muito_baixo'] = base_perfil['Menos de 1/2 salário mínimo'] / base_perfil['Total_gasto']

base_perfil['prop_gasto_baixo'] = base_perfil['1/2 a menos de 1 salário mínimo'] / base_perfil['Total_gasto']

base_perfil['prop_gasto_medio_baixo'] = base_perfil['1 a menos de 2 salários mínimos'] / base_perfil['Total_gasto']

base_perfil['prop_gasto_medio'] = base_perfil['2 a menos de 4 salários mínimos'] / base_perfil['Total_gasto']

base_perfil['prop_gasto_alto'] = base_perfil['4 ou mais salários mínimos'] / base_perfil['Total_gasto']
```

```python
# Guardar identificadores
ids = base_perfil[['UF', 'Ano']].copy()

# Base apenas para clusterização
base_s_anouf = base_perfil.drop(columns=['UF', 'Ano'])
base_s_anouf= base_perfil.filter(like='prop_')
base_s_anouf

```

```python
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
```

```python
# Seleciona apenas colunas de proporção


perfil_cluster = base_s_anouf.select_dtypes(include= ['float64', 'int64']).copy()
colunas = perfil_cluster.columns
indice = perfil_cluster.index

normalizar= StandardScaler()
perfil_cluster_normalizado = normalizar.fit_transform(perfil_cluster)
display ()


```

```python
import matplotlib.pyplot as plt
```

```python
import matplotlib.pyplot as plt
from sklearn.metrics import silhouette_score

inertia = []
K = range(2, 11)

for k in K:
    kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
    kmeans.fit(perfil_cluster_normalizado)
    inertia.append(kmeans.inertia_)


#cotovelo
plt.figure(figsize=(8,4))
plt.plot(K, inertia, 'bo-')
plt.xlabel('Número de Clusters (k)')
plt.ylabel('Inércia')
plt.title('Método do Cotovelo (Elbow)')
plt.show()
```

```python
#%% Cluster K-means
perfil_cluster_normalizado = pd.DataFrame(
    perfil_cluster_normalizado,
    columns=perfil_cluster.columns,
    index=perfil_cluster.index
)


kmeans_final = KMeans(n_clusters = 3, init = 'k-means++', random_state=100).fit(perfil_cluster_normalizado)

# Gerando a variável para identificarmos os clusters gerados
kmeans_clusters = kmeans_final.labels_
base_perfil['Cluster'] = kmeans_clusters
base_perfil['Cluster'] = base_perfil['Cluster'].astype('category')

cent_finais= pd.DataFrame(kmeans_final.cluster_centers_)
cent_finais.columns= perfil_cluster.columns
cent_finais.index.name ="cluster"
cent_finais


```

```python

!pip install pingouin
import pingouin as pg


colunas_para_testar = [
    'prop_pessoal',
    'prop_profissional',
    'prop_hotel',
    'prop_pousada',
    'prop_casa',
    'prop_imovel_proprio',
    'prop_airbnb',
    'prop_outro_hosp',
    'prop_aviao',
    'prop_carro',
    'prop_moto',
    'prop_onibus_exc',
    'prop_onibus_linha',
    'prop_van',
    'prop_outro_transp',
    'prop_gasto_muito_baixo',
    'prop_gasto_baixo',
    'prop_gasto_medio_baixo',
    'prop_gasto_medio',
    'prop_gasto_alto'
]

resultados = []

for coluna in colunas_para_testar:

    try:
        aov = pg.anova(
            dv=coluna,
            between='Cluster',
            data=base_perfil,
            detailed=True
        )

        f_valor = aov.loc[0, 'F']
        p_valor = aov.loc[0, 'p_unc']

        resultados.append({
            'Variavel': coluna,
            'F_Estatistica': f_valor,
            'p_valor': p_valor
        })

    except Exception as e:
        print(f"Erro na variável {coluna}: {e}")

tabela_rank = pd.DataFrame(resultados)

if not tabela_rank.empty:
    tabela_rank = tabela_rank.sort_values(by='F_Estatistica', ascending=False)
    tabela_rank = tabela_rank.reset_index(drop=True)

tabela_rank = tabela_rank.round(4)

print(tabela_rank)
tabela_rank.to_csv('/content/tabela_rank_final.csv', index=False)
```
